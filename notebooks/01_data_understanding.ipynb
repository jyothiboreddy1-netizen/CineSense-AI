```python
import pandas as pd
import numpy as np

movies = pd.read_csv("movies.csv")
ratings = pd.read_csv("ratings.csv")
tags = pd.read_csv("tags.csv")
links = pd.read_csv("links.csv")

print("Movies shape:", movies.shape)
print("Ratings shape:", ratings.shape)
print("Tags shape:", tags.shape)
print("Links shape:", links.shape)
```

```python
print("MOVIES DATASET")
display(movies.head())

print("\nRATINGS DATASET")
display(ratings.head())

print("\nTAGS DATASET")
display(tags.head())

print("\nLINKS DATASET")
display(links.head())
```

```python
# Check basic information about each dataset

datasets = {
    "Movies": movies,
    "Ratings": ratings,
    "Tags": tags,
    "Links": links
}

for name, df in datasets.items():
    print("=" * 50)
    print(name.upper(), "DATASET")
    print("=" * 50)

    print("\nShape:", df.shape)

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:", df.duplicated().sum())

    print()
```

```python
print("Number of unique users:", ratings["userId"].nunique())
print("Number of unique rated movies:", ratings["movieId"].nunique())

print("\nRating distribution:")
print(ratings["rating"].value_counts().sort_index())

print("\nRating statistics:")
display(ratings["rating"].describe())
```

```python
# Create copies to preserve original data
movies_clean = movies.copy()
ratings_clean = ratings.copy()
tags_clean = tags.copy()
links_clean = links.copy()

# Extract release year from movie title
movies_clean["year"] = movies_clean["title"].str.extract(r"\((\d{4})\)$")

# Remove year from title to create a clean movie title
movies_clean["clean_title"] = movies_clean["title"].str.replace(
    r"\s*\(\d{4}\)$", "", regex=True
)

# Convert year to numeric
movies_clean["year"] = pd.to_numeric(
    movies_clean["year"], errors="coerce"
)

# Convert Unix timestamps to readable datetime
ratings_clean["datetime"] = pd.to_datetime(
    ratings_clean["timestamp"], unit="s"
)

tags_clean["datetime"] = pd.to_datetime(
    tags_clean["timestamp"], unit="s"
)

# Preview processed data
print("PROCESSED MOVIES")
display(movies_clean.head())

print("\nPROCESSED RATINGS")
display(ratings_clean.head())

print("\nPROCESSED TAGS")
display(tags_clean.head())
```

```python
# Split genres into Python lists
movies_clean["genre_list"] = movies_clean["genres"].str.split("|")

# Create a lowercase version of tags for consistent NLP processing
tags_clean["tag_clean"] = (
    tags_clean["tag"]
    .str.lower()
    .str.strip()
)

# Combine all tags belonging to the same movie
movie_tags = (
    tags_clean.groupby("movieId")["tag_clean"]
    .apply(lambda x: " ".join(x))
    .reset_index()
)

movie_tags.rename(columns={"tag_clean": "combined_tags"}, inplace=True)

# Merge combined tags with movie information
movies_enriched = movies_clean.merge(
    movie_tags,
    on="movieId",
    how="left"
)

# Movies without user tags get an empty string
movies_enriched["combined_tags"] = movies_enriched["combined_tags"].fillna("")

print("ENRICHED MOVIE DATA")
display(
    movies_enriched[
        ["movieId", "clean_title", "year", "genres",
         "genre_list", "combined_tags"]
    ].head(10)
)

print("\nShape:", movies_enriched.shape)
```

```python
# Calculate rating statistics for each movie
rating_stats = (
    ratings_clean.groupby("movieId")
    .agg(
        rating_count=("rating", "count"),
        average_rating=("rating", "mean")
    )
    .reset_index()
)

# Merge rating statistics with enriched movie data
movies_final = movies_enriched.merge(
    rating_stats,
    on="movieId",
    how="left"
)

# Fill missing values for movies with no ratings
movies_final["rating_count"] = movies_final["rating_count"].fillna(0).astype(int)
movies_final["average_rating"] = movies_final["average_rating"].fillna(0)

# Round average rating
movies_final["average_rating"] = movies_final["average_rating"].round(2)

print("FINAL MOVIE DATASET")
display(
    movies_final[
        [
            "movieId",
            "clean_title",
            "year",
            "genres",
            "combined_tags",
            "rating_count",
            "average_rating"
        ]
    ].head(10)
)

print("\nFinal Shape:", movies_final.shape)
```

```python
import os

# Create output folder
os.makedirs("processed_data", exist_ok=True)

# Save processed datasets
movies_final.to_csv(
    "processed_data/movies_final.csv",
    index=False
)

ratings_clean.to_csv(
    "processed_data/ratings_clean.csv",
    index=False
)

tags_clean.to_csv(
    "processed_data/tags_clean.csv",
    index=False
)

print("Processed datasets saved successfully!")
```
